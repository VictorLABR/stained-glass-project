# Probability - Task

```python
########## IMPORTS BLOCK ##########

import numpy as np
import matplotlib.pyplot as plt

from scipy import stats
from scipy.stats import norm, t
```

1. A machine packs dog food into 15 kg packages, but each package produced ends up having a random weight, with Normal distribution, with a mean of 15 kg, but with a standard deviation of 0,1 kg. Find the probability that a randomly selected package from this production is greater than 15,2 kg.

```python
########## VARIABLES DEFFINITION ##########

w = 15
σ = 0.1
w_threshold = 15.2

########## PROB CALCULATION ##########

p = 1 - norm.cdf(w_threshold, loc = w, scale = σ)

########## RESULTS EXHIBITION ##########

print("The probability of a package being greater than 15.2 kg is {:.2f}%".format(p * 100))
```

2. From the machine described above, obtain the 95% percentile for the weight of the packages. In other words, the value such that the probability of a package weighing less than this is 95%.

```python
percentile95 = norm.ppf(0.95, loc = w, scale = σ)

########## RESULTS EXHIBITION ##########

print("The 95% percentile for packaging weight is {:.2f} kg".format(percentile95))
```

3. Suppose two random variables $Y_1$ and $Y_2$, the first with a Standard Normal distribution (mean=0, variance=1), the second with a *t-studdent* distribution with 5 degrees of freedom (also standard). Calculate, for each one, the probability of being less than each of the values contained in the object $Y$ in the cell below:

```python
Y = np.linspace(-6, 6, 200)

########## PROBS CALCULATION ##########

prob_Y_normal = norm.cdf(Y, loc = 0, scale = 1)

prob_Y_t_student = t.cdf(Y, df = 5)
```

4. Make a graphical visualization of the data from the previous item.

```python
########## GRAPHING THE CURVES ##########

plt.figure(figsize=(10, 6))
plt.plot(Y, prob_Y_normal, label = 'Normal')
plt.plot(Y, prob_Y_t_student, label = 't-Student (df = 5)')
plt.xlabel('Y')
plt.ylabel('Probability')
plt.title('Cumulative Distribution Function (CDF) - Normal vs t-Student Distributions')
plt.legend()
plt.grid(True)
plt.show()
```

5. Suppose two random variables $Y_1$ and $Y_2$, the first with a Standard Normal distribution (mean=0, variance=1), the second with a *t-studdent* distribution with 5 degrees of freedom (also standard). Calculate, for each one, the percentile (the inverse of *CDF*) for each of the values of the *P* object below:

```python
Y=np.linspace(0, 1, 20)

########## PROBS CALCULATION ##########

percentile_Y_normal = norm.ppf(Y, loc=0, scale=1)

percentile_Y_t_student = t.ppf(Y, df=5)
```

6. Make a graph with the data from the previous item.

```python
########## GRAPHING THE CURVES ##########

plt.figure(figsize=(10, 6))
plt.plot(Y, percentile_Y_normal, label = 'Normal')
plt.plot(Y, percentile_Y_t_student, label = 't-Student (df = 5)')
plt.xlabel('Y')
plt.ylabel('Percentile')
plt.title('Percentiles (PPF) - Normal vs t-Student Distributions')
plt.legend()
plt.grid(True)
plt.show()
```
